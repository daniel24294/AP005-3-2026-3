Este bloque de código importa la librería pandas.

```python
import pandas as pd
```

Este bloque de código crea un diccionario llamado data con información sobre nombres, edades y género.

```python
data = {"name": ["Sally", "Jhon", "Kim"], "age": [4, 6 ,7], "gender": ["female", "male", "male"]}
```

Este bloque de código convierte el diccionario data en un DataFrame.

```python
pd.DataFrame(data)
```

Este bloque de código crea un DataFrame asignando índices personalizados a cada fila.

```python
pd.DataFrame(data, index = ["Kid1", "Kid2", "Kid3"])
```
