# PRÁCTICA 1. NEURONA ARTIFICIAL SIMPLE: `neuron1`

## 1. Descripción y modelo matemático

En esta práctica se implementa una neurona artificial de una sola entrada y un único peso, con el propósito de observar cómo un parámetro puede ajustarse mediante el entrenamiento para aproximar una relación entre entradas y salidas esperadas.

Los datos de entrenamiento son:

- **Entradas:** $(x=[1,2,3,4])$
- **Salidas esperadas:** $(t=[2,4,6,8])$

Estos valores siguen la relación lineal $(t=2x)$. Por tanto, el objetivo es que la neurona aprenda a aproximar esta relación modificando su peso inicial.

El modelo utilizado es:

$
\boxed{y=wx}
$

Donde $(x)$ representa la entrada, $(w)$ es el peso ajustable y $(y)$ es la predicción de la neurona. En este modelo no se incorpora un sesgo.

## 2. Variables y parámetros del entrenamiento

| Variable | Valor | Función |
|---|---|---|
| `inputs` | [1, 2, 3, 4] | Entradas de entrenamiento |
| `targets` | [2, 4, 6, 8] | Salidas esperadas |
| `w` | 0.1 | Peso inicial |
| `learning_rate` | 0.1 | Tasa de aprendizaje |
| `epochs` | 30 | Número de épocas |
| `pause_time` | 1 s | Pausa entre épocas |

El entrenamiento comienza con \(w=0.1\), un valor alejado del peso esperado \(w=2\). La tasa de aprendizaje determina qué fracción del error promedio se utiliza para modificar el peso en cada época.

La pausa de un segundo permite observar la evolución del programa, pero no interviene en los cálculos del aprendizaje.

## 3. Proceso de entrenamiento

En cada época, el programa ejecuta cuatro operaciones: calcula las predicciones, obtiene los errores individuales, calcula su promedio y actualiza el peso.

**a. Predicción**

La salida se calcula multiplicando cada entrada por el peso actual:

$y_i = w x_i$

**b. Cálculo del error**

El error individual corresponde a la diferencia entre el valor esperado y la predicción:

$e_i=t_i-y_i$


Un error positivo indica que la predicción es inferior al valor esperado, mientras que uno negativo indica que lo supera.

**c. Error promedio**

El programa calcula la media aritmética de los errores:

$
\text{cost}=\frac{1}{n}\sum_{i=1}^{n}(t_i-y_i)
$

En esta práctica, `cost` representa el error promedio con signo, no el error cuadrático.

**d. Actualización del peso**

El peso se modifica mediante la siguiente regla:

$
\boxed{w_{\text{nuevo}}=w_{\text{anterior}}+
\eta\cdot\text{cost}}
$

Donde $\eta\$ es la tasa de aprendizaje. Al utilizar una tasa de 0.1, se suma al peso una décima parte del error promedio.

Por ejemplo, en la primera época, las predicciones iniciales son $([0.1,0.2,0.3,0.4])$, mientras que los errores son $([1.9,3.8,5.7,7.6])$. El error promedio es:

$
\text{cost}=\frac{1.9+3.8+5.7+7.6}{4}=4.75
$

Por tanto, el nuevo peso es:

$
w_{\text{nuevo}}=0.1+(0.1)(4.75)=0.575
$

Este procedimiento se repite durante las 30 épocas. En cada repetición, las predicciones se calculan con el peso actualizado.

## 4. Resultados, pruebas e interpretación

### 4.1. Evolución del entrenamiento

Durante el entrenamiento se muestran las entradas, los valores esperados, las predicciones, los errores y el peso antes y después de cada actualización.

Se espera que el peso se aproxime progresivamente a \(2\), ya que la relación presente en los datos es \(t=2x\). En consecuencia, las predicciones deberían acercarse a las salidas esperadas a medida que avanza el entrenamiento.

La siguiente tabla resume el comportamiento inicial y el resultado esperado:

| Etapa | Peso aproximado | Interpretación |
|---|---:|---|
| Inicio | 0.1000 | Predicciones alejadas de los valores esperados |
| Primera época | 0.5750 | El peso aumenta al obtenerse un error promedio positivo |
| Final del entrenamiento | 1.9997 | El peso se aproxima a 2 |

*Los valores de la tabla corresponden al cálculo teórico con los parámetros definidos en el programa.*

### 4.2. Prueba con nuevos datos

Después del entrenamiento se utilizan las entradas \(x=[5,6]\), que no participaron en el ajuste del peso. Sus salidas esperadas son \(t=[10,12]\).

| Entrada | Salida esperada | Predicción ideal |
|---:|---:|---:|
| 5 | 10 | 10 |
| 6 | 12 | 12 |

La comparación permite comprobar si la neurona aprendió la relación lineal y si puede aplicarla a entradas diferentes de las utilizadas durante el entrenamiento. Las predicciones reales pueden presentar una pequeña diferencia respecto a los valores ideales.

### 4.3. Representación gráfica

Para complementar los resultados, se utilizarán dos gráficas:

- **Evolución del peso:** permite observar cómo cambia $(w)$ durante las épocas y cómo se aproxima al valor de referencia $(w=2)$.
- **Predicciones frente a valores esperados:** permite comparar las salidas de la neurona con la relación ideal $(y=2x)$.

Estas representaciones facilitan la identificación visual del aprendizaje y de las diferencias entre las predicciones obtenidas y los valores deseados.

### Conclusión

La práctica permite comprender el funcionamiento básico de una neurona artificial con salida lineal. A través de la predicción, el cálculo del error promedio y la actualización del peso, se observa cómo un parámetro inicialmente inadecuado puede ajustarse para aproximar una relación entre entradas y salidas. También se comprueba la importancia de evaluar el modelo con datos que no participaron en el entrenamiento. 

```python
import time

# Datos de entrenamiento
inputs = [1, 2, 3, 4]
targets = [2, 4, 6, 8]

# Peso inicial de la neurona
w = 0.1

# Tasa de aprendizaje
learning_rate = 0.1

# Numero total de epocas
epochs = 30

# Tiempo de pausa entre epocas, en segundos
pause_time = 1


# Funcion de prediccion
# Modelo de la neurona: y = w*x
def predict(i):
    return w * i


# Entrenamiento
for epoch in range(epochs):

    # Prediccion para todos los datos de entrenamiento
    predictions = [predict(i) for i in inputs]

    # Calculo del error de cada dato
    # error = target - prediction
    errors = [
        t - p
        for p, t in zip(predictions, targets)
    ]

    # Error promedio de la epoca
    cost = sum(errors) / len(targets)

    # Mostrar los resultados de la epoca
    print("\n" + "=" * 60)
    print(f"EPOCA {epoch + 1}")
    print("=" * 60)

    for i, t, p, e in zip(
        inputs,
        targets,
        predictions,
        errors
    ):
        print(
            f"Entrada: {i:>2} | "
            f"Target: {t:>2} | "
            f"Pred: {p:>8.4f} | "
            f"Error: {e:>8.4f}"
        )

    print("-" * 60)
    print(f"Peso antes de actualizar: {w:.10f}")
    print(f"Error promedio: {cost: .6f}")

    # Actualizacion del peso
    w += learning_rate * cost

    print(f"Peso despues de actualizar: {w:.10f}")

    # Esperar antes de mostrar la siguiente epoca
    time.sleep(pause_time)


# Datos de prueba
test_inputs = [5, 6]
test_targets = [10, 12]

print("\n" + "=" * 60)
print("PRUEBA FINAL")
print("=" * 60)

# Predicciones con datos que no participaron
# en el entrenamiento
test_predictions = [predict(i) for i in test_inputs]

for i, t, p in zip(
    test_inputs,
    test_targets,
    test_predictions
):
    print(
        f"Entrada: {i:>2} | "
        f"Target: {t:>2} | "
        f"Prediccion: {p:>8.4f}"
    )
```
