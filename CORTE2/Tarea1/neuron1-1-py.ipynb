# PRÁCTICA 2. NEURONA ARTIFICIAL SIMPLE: `neuron1.1` 

## 1. Descripción y Fundamento Matemático
(Optimización por Gradientes y Error Cuadrático)
En esta segunda práctica se implementa una neurona artificial de una sola entrada y un único peso sináptico. El propósito principal es observar cómo la neurona ajusta sus parámetros mediante el cálculo analítico de derivadas y una función de costo basada en el error cuadrático.

Los datos de entrenamiento son:
- **Entradas ($x$):** $[1, 2, 3, 4]$
- **Salidas esperadas ($t$):** $[2, 4, 6, 8]$

Estos valores siguen la relación lineal $t = 2x$. La neurona parte de un peso inicial arbitrario ($w = 0.1$) y debe modificarlo iterativamente para aproximarse a la relación funcional de los datos.

El modelo matemático utilizado es:
$$\boxed{y = w \cdot x}$$
Donde $x$ es la entrada, $w$ es el peso sináptico ajustable e $y$ es la predicción calculada por el modelo.

---

## 2. Variables y Parámetros del Entrenamiento

| Variable | Valor | Función en el Algoritmo |
| :--- | :---: | :--- |
| `inputs` | `[1, 2, 3, 4]` | Arreglo con los valores de entrada para el entrenamiento. |
| `targets` | `[2, 4, 6, 8]` | Arreglo con las salidas ideales o esperadas. |
| `w` | `0.1` | Valor inicial del peso sináptico. |
| `epochs` | `10` | Número total de épocas o iteraciones de entrenamiento. |
| `learning_rate` | `0.1` | Tasa de aprendizaje que determina la magnitud de la actualización del peso. |
| `pause_time` | `1` | Tiempo de pausa (en segundos) para la visualización didáctica entre épocas. |

---

## 3. Proceso de Entrenamiento y Fundamento de las Derivadas

Durante el bucle de entrenamiento, el programa ejecuta de manera secuencial las siguientes etapas matemáticas en cada época:

### a. Predicción
La salida estimada se calcula multiplicando la entrada por el peso actual:
$$p_i = w \cdot x_i$$

### b. Error Cuadrático y Costo Promedio
El error de cada muestra se evalúa elevando al cuadrado la diferencia entre la predicción y el valor esperado, lo que penaliza de forma no lineal las desviaciones grandes:
$$E_i = (p_i - t_i)^2$$
El costo promedio ($J$) de la época es la media de estos errores cuadráticos:
$$J = \frac{1}{n} \sum_{i=1}^{n} (p_i - t_i)^2$$

### c. Derivadas y Gradientes
Para aplicar el descenso por gradiente, se calculan las derivadas parciales:
1. **Derivada del error respecto a la salida ($\frac{\partial E}{\partial y}$):**
   $$\frac{\partial E_i}{\partial y_i} = 2(p_i - t_i)$$
2. **Derivada del error respecto al peso ($\frac{\partial E}{\partial w}$):** aplicando la regla de la cadena considerando que $\frac{\partial y}{\partial w} = x_i$:
   $$\frac{\partial E_i}{\partial w} = 2(p_i - t_i) \cdot x_i$$
3. **Gradiente promedio ($\nabla J$):** Corresponde a la media aritmética de las derivadas individuales de todos los datos:
   $$\text{gradient} = \frac{1}{n} \sum_{i=1}^{n} \frac{\partial E_i}{\partial w}$$

### d. Regla de Actualización
El peso se actualiza restando una fracción del gradiente dictada por la tasa de aprendizaje ($\eta$):
$$\boxed{w_{\text{nuevo}} = w_{\text{anterior}} - \eta \cdot \text{gradient}}$$

---

## 4. Interpretación de Resultados y Ejecución Real

### 4.1. Evolución del Entrenamiento
Durante la ejecución de las 10 épocas, el sistema muestra en tiempo real cómo desciende el costo promedio y cómo el peso $w$ se modifica a partir del valor inicial de $0.1$. Debido a la tasa de aprendizaje y al número finito de épocas, el peso experimenta oscilaciones iniciales controladas hasta estabilizarse cerca del valor óptimo (por ejemplo, llegando a un valor aproximado de $1.9981$ al concluir la décima época).
### 4.2. Evolución del Entrenamiento
| Característica | Práctica Anterior (`neuron1.py`) | Práctica Actual (`neuron1.1.py`) |
| :--- | :--- | :--- |
| **Función de Error** | Error lineal directo con signo: $e = t - p$ | **Error cuadrático individual:** $E = (p - t)^2$ |
| **Naturaleza del Error** | Puede cancelarse si hay errores positivos y negativos mixtos. | Siempre positivo o cero; penaliza de forma exponencial las desviaciones grandes. |
| **Métrica de Costo** | Promedio simple de errores con signo ($\text{cost}$). | **Costo promedio cuadrático ($J$):** Media aritmética de los errores al cuadrado. |
| **Regla de Actualización** | Ajuste directo basado en la tasa y el error promedio: $w \leftarrow w + \eta \cdot \text{cost}$. | **Descenso por gradiente:** Derivada analítica del error respecto al peso: $w \leftarrow w - \eta \cdot \text{gradient}$. |

```python
import time

# Datos de entrenamiento
inputs = [1, 2, 3, 4]
targets = [2, 4, 6, 8]

# Peso inicial
w = 0.1

# Numero de epocas
epochs = 10

# Tasa de aprendizaje
learning_rate = 0.1

# Tiempo de pausa entre epocas
pause_time = 1


# Funcion de prediccion
# Modelo: y = w*x
def predict(i):
    return w * i


# Entrenamiento
for epoch in range(epochs):

    # Predicciones para todos los datos
    predictions = [predict(i) for i in inputs]

    # Error cuadratico para cada dato
    errors = [
        (p - t)**2
        for p, t in zip(predictions, targets)
    ]

    # Costo promedio
    cost = sum(errors) / len(targets)

    # Derivada del error con respecto a la salida
    errors_d = [
        2 * (p - t)
        for p, t in zip(predictions, targets)
    ]

    # Derivada del error con respecto al peso
    weight_d = [
        e * i
        for e, i in zip(errors_d, inputs)
    ]

    # Mostrar informacion de la epoca
    print("\n" + "=" * 65)
    print(f"EPOCA {epoch + 1}")
    print("=" * 65)

    for i, t, p, e, ed, wd in zip(
        inputs,
        targets,
        predictions,
        errors,
        errors_d,
        weight_d
    ):
        print(
            f"Entrada: {i:>2} | "
            f"Target: {t:>2} | "
            f"Pred: {p:>8.4f} | "
            f"Error^2: {e:>8.4f} | "
            f"dE/dy: {ed:>8.4f} | "
            f"dE/dw: {wd:>8.4f}"
        )

    print("-" * 65)
    print(f"Peso antes de actualizar: {w:.10f}")
    print(f"Costo promedio: {cost:.6f}")

    # Gradiente promedio respecto al peso
    gradient = sum(weight_d) / len(weight_d)

    print(f"Gradiente promedio: {gradient:.6f}")

    # Actualizacion del peso
    w -= learning_rate * gradient

    print(f"Peso despues de actualizar: {w:.10f}")

    # Pausa didactica entre epocas
    time.sleep(pause_time)


# Datos de prueba
test_inputs = [5, 6]
test_targets = [10, 12]

print("\n" + "=" * 65)
print("PRUEBA FINAL")
print("=" * 65)

test_predictions = [
    predict(i)
    for i in test_inputs
]

for i, t, p in zip(
    test_inputs,
    test_targets,
    test_predictions
):
    print(
        f"Entrada: {i:>2} | "
        f"Target: {t:>2} | "
        f"Prediccion: {p:>8.4f}"
    )
```
